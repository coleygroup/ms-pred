```python
def weighted_group_average(inp_df):
    # rm the current "avg" row
    df_filtered = inp_df[inp_df.iloc[:, 0] != 'avg']

    # Filter positive and negative rows
    positive_rows = df_filtered[df_filtered.iloc[:, 0].str.endswith('+')]
    negative_rows = df_filtered[df_filtered.iloc[:, 0].str.endswith('-')]
    metfrag_rows = df_filtered[df_filtered.iloc[:, 0].str.endswith(('[M+H]+', '[M-H]-', '[M+Na]+', '[M+K]+', '[M+Cl]-', '[M+H3N+H]+'))]

    def weighted_avg(df):
        weights = df['num_examples']
        result = {}
        for col in df.select_dtypes(include=[np.number]).columns:
            result[col] = np.average(df[col], weights=weights)
        return result
    
    # List of all original columns
    columns = inp_df.columns

    # Create new row dictionaries and include labels for 'Unnamed: 0'
    def create_row_dict(weighted_dict, label, all_columns):
        # For each column in the original DataFrame, use the weighted value if available, otherwise NaN.
        row = {col: weighted_dict[col] if col in weighted_dict else np.nan for col in all_columns}
        row[columns[0]] = label
        return row

    
    # Calculate weighted averages for each group and create new rows
    # 6, 3/4, 9/10 are the current experiment setup (scaffold split does not have [M-H-CO2]- in test set)
    weighted_df = []
    if len(positive_rows) == 6:
        weighted_avg_positive = weighted_avg(positive_rows)
        wp_row = create_row_dict(weighted_avg_positive, 'pos_avg', columns)
        weighted_df.append(wp_row)
    if len(negative_rows) == 3 or len(negative_rows) == 4: 
        weighted_avg_negative = weighted_avg(negative_rows)
        wn_row = create_row_dict(weighted_avg_negative, 'neg_avg', columns)
        weighted_df.append(wn_row)
    if len(df_filtered) == 9 or len(df_filtered) == 10:
        weighted_avg_general  = weighted_avg(df_filtered)
        wg_row = create_row_dict(weighted_avg_general,  'avg',  columns)
        weighted_df.append(wg_row)
    if len(metfrag_rows) == 6:
        weighted_avg_metfrag  = weighted_avg(metfrag_rows)
        mf_row = create_row_dict(weighted_avg_metfrag,  'metfrag',  columns)
        weighted_df.append(mf_row)

    # Create a DataFrame for these new rows
    weighted_df = pd.DataFrame(weighted_df)

    # Concatenate the new rows with the original DataFrame
    out_df = pd.concat([df_filtered, weighted_df], ignore_index=True)
    
    return out_df
```

```python
names = [
    "CFM-ID",
    "MetFrag",
    "GrAFF-MS",
    "MassFormer",
    "FraGNNet",
    "ICEBERG 1.0",
    "ICEBERG 2.0",
]

all_adducts = ['[M+H]+', '[M-H2O+H]+', '[M-H4O2+H]+', '[M+Na]+', '[M+K]+', '[M+H3N+H]+', # pos
               '[M-H]-', '[M+Cl]-', '[M-H-H2O]-', '[M-H-CO2]-'] # neg

dataset_to_res = {}
for dataset_name in dataset_names:
    for split_name in ['split_1', 'scaffold_1']:
        tsv_files = defaultdict(lambda : [])
        for seed in [1,2,3]:    
            results_files = [
                f"results/cfm_id_{dataset_name}/{split_name}/preds/pred_eval_grouped_ion_type.tsv",
                None,
                f"results/graff_ms_baseline_{dataset_name}/{split_name}_rnd{seed}/preds/pred_eval_grouped_ion_type.tsv",
                f"results/massformer_baseline_{dataset_name}/{split_name}_rnd{seed}/preds/pred_eval_grouped_ion_type.tsv",
                f"results/fragnet_preprint_{dataset_name}/{split_name}_rnd{seed}/preds/pred_eval_grouped_ion_type.tsv",
                f"../ms-pred-legacy/results/dag_inten_{dataset_name}/{split_name}_rnd{seed}/preds/pred_eval_grouped_ion_type.tsv", # old ICEBERG 1.0 (with Analy Chem'24)
                f"results/iceberg_{dataset_name}/{split_name}_rnd{seed}/preds/pred_eval_grouped_ion_type.tsv",
            ]
            for i, j in zip(names, results_files):
                if j and Path(j).exists():
                    cos_df = pd.read_csv(j, sep='\t')
                else:
                    cos_df = None
                if j:
                    retrieval_path_entr = Path(j).parent / f'../retrieval_{dataset_name}_{split_name}_50/rerank_eval_grouped_ion_entropy.tsv'
                    retrieval_path_cos = Path(j).parent / f'../retrieval_{dataset_name}_{split_name}_50/rerank_eval_grouped_ion_cos.tsv'
                    if retrieval_path_entr.exists():
                        retrieval_df = pd.read_csv(retrieval_path_entr, sep='\t')
                    elif retrieval_path_cos.exists():
                        retrieval_df = pd.read_csv(retrieval_path_cos, sep='\t')
                    else:
                        retrieval_df = None
                elif i == 'MetFrag':
                    retrieval_path = f'results/metfrag_{dataset_name}/{split_name}/retrieval/rerank_eval_grouped_ion_met_frag.tsv'
                    retrieval_df = pd.read_csv(retrieval_path, sep='\t')
                else:
                    retrieval_df = None

                if cos_df is not None:
                    cos_df = weighted_group_average(cos_df)
                if retrieval_df is not None:
                    retrieval_df = weighted_group_average(retrieval_df)
                
                concat_df = pd.concat((cos_df, retrieval_df), axis=1)
                tsv_files[i].append(concat_df)

        dataset_to_res[f"{split_name}_{dataset_name}"] = tsv_files
```

```python
adduct_to_df = {}
for adduct_type in all_adducts + ['pos_avg', 'neg_avg', 'avg', 'metfrag']:
    out_df = []
    for dataset_name, tsv_files in dataset_to_res.items():
        for k, v in tsv_files.items():
            interested_rows = [vv[vv.iloc[:, 0] == adduct_type] for vv in v]
            if any([len(row) == 0 for row in interested_rows]):
                continue
            if k == 'CFM-ID' and adduct_type != '[M+H]+': # only evaluated [M+H]+ for CFM-ID
                continue
            def get_val(key):
                vals = []
                for row in interested_rows:
                    if key in row:
                        vals.append(row[key].item())
                return vals
            new_entry = {
                "Method": k,
                "Cosine sim.": np.mean(get_val('cos_sim')),
                "Cosine sim. 95%": 1.96 * sem(get_val('cos_sim')),

                "Cosine sim. (no PEP)": np.mean(get_val('cos_sim_zero_pep')),
                "Cosine sim. 95% (no PEP)": 1.96 * sem(get_val('cos_sim_zero_pep')),

                "Entropy sim.": np.mean(get_val('entropy_sim')),
                "Entropy sim. 95%": 1.96 * sem(get_val('entropy_sim')),

                "Coverage": np.mean(get_val('coverage')),
                "Coverage 95%": 1.96 * sem(get_val('coverage')),

                # The way of computing validity in this code base is outdated
                #"Valid": np.mean(get_val('frac_valid')),
                #"Valid 95%": 1.96 * sem(get_val('frac_valid')),

                "Dataset": dataset_name,
            }
            new_entry.update({
                f"Top_{i} Retrieval": np.mean(get_val(f"top_{i}")) for i in range(1, 11)
            })
            new_entry.update({
                f"Top_{i} Retrieval 95%": 1.96 * sem(get_val(f"top_{i}")) for i in range(1, 11)
            })
            out_df.append(new_entry)
    out_df = pd.DataFrame(out_df)

    # Replace nan with 0 
    out_df = out_df.fillna(0)
    
    adduct_to_df[adduct_type] = out_df
```

```python
for adduct_type, out_df in adduct_to_df.items():
    # rewrite cosine sim column with f"cos sim \pm cosine sem using list comprehension
    out_df["Cosine sim. orig"] = out_df["Cosine sim."] 

    out_df["Cosine sim."] = [
        rf"${i:.3f} \pm {j:.3f}$"
        for i, j in zip(out_df["Cosine sim."], out_df["Cosine sim. 95%"])
    ]
    # Same for entropy
    out_df["Entropy sim."] = [
        rf"${i:.3f} \pm {j:.3f}$"
        for i, j in zip(out_df["Entropy sim."], out_df["Entropy sim. 95%"])
    ]
    # Rewrite coverage column with f"coverage \pm coverage sem using list comprehension
    out_df["Coverage"] = [
        rf"${i:.3f} \pm {j:.3f}$"
        for i, j in zip(out_df["Coverage"], out_df["Coverage 95%"])
    ]
    # Rewrite all top_K retrieval accuracy columns
    for i in range(1, 11):
    #     out_df[f"Top_{i} Retrieval orig"] = out_df[f"Top_{i} Retrieval"]
        out_df[f"Top-{i}"] = [
            rf"${i:.3f} \pm {j:.3f}$"
            for i, j in zip(out_df[f"Top_{i} Retrieval"], out_df[f"Top_{i} Retrieval 95%"])
        ]
```

```python
## Define Cosine sim bar plots

def get_metric_points(active_adduct, dataset, metric_col="cos_sim"):
    method_to_points = {}
    tsv_files = dataset_to_res.get(dataset, {})
    for method in names:
        vals = []
        for run_df in tsv_files.get(method, []):
            if run_df is None or len(run_df) == 0 or metric_col not in run_df.columns:
                continue
            rows = run_df[run_df.iloc[:, 0] == active_adduct]
            if rows.empty:
                continue
            val = rows[metric_col].iloc[0]
            if pd.notna(val) and np.isfinite(val) and val > 0:
                vals.append(float(val))
        if method in {"CFM-ID", "MetFrag"} and vals:
            vals = list(dict.fromkeys(vals))
        method_to_points[method] = vals
    return method_to_points


def format_n_counts(n_counts):
    return "; ".join([f"{method}={n}" for method, n in n_counts.items() if n > 0])


def format_n_counts_wrapped(prefix, n_counts, per_line=3):
    items = [f"{method}={n}" for method, n in n_counts.items() if n > 0]
    if not items:
        return f"{prefix}: n/a"
    chunks = [items[i : i + per_line] for i in range(0, len(items), per_line)]
    lines = [f"{prefix}: {', '.join(chunks[0])}"]
    lines.extend([f"  {', '.join(chunk)}" for chunk in chunks[1:]])
    return "\n".join(lines)


def overlay_metric_points(ax, method_names, valid_indices, method_to_points):
    plotted_methods = np.array(method_names)[valid_indices]
    for x_pos, method in enumerate(plotted_methods):
        points = method_to_points.get(method, [])
        if not points:
            continue
        offsets = np.linspace(-0.12, 0.12, len(points)) if len(points) > 1 else [0]
        facecolors = [(1, 1, 1, 0.5)] * len(points)
        ax.scatter(
            np.full(len(points), x_pos) + offsets,
            points,
            s=5,
            marker="o",
            facecolors=facecolors,
            edgecolors="black",
            linewidth=0.35,
            zorder=5,
            clip_on=False,
        )


def draw_visible_ci(ax, x, y, ci):
    ax.errorbar(
        x,
        y,
        yerr=np.nan_to_num(np.asarray(ci, dtype=float), nan=0.0),
        fmt="none",
        ecolor="black",
        elinewidth=0.65,
        capsize=2.0,
        capthick=0.65,
        zorder=7,
    )


def annotate_n_counts(ax, x_pos, y_vals, methods, n_counts):
    for x, y, method in zip(x_pos, y_vals, methods):
        n = n_counts.get(method, 0)
        if n <= 0:
            continue
        ax.text(x, min(float(y) + 0.025, 0.94), f"n={n}", ha="center", va="bottom", fontsize=5, zorder=10)


dataset = "split_1_nist20"
dataset_sub = out_df[["Method", "Cosine sim. orig", "Cosine sim. 95%", "Dataset"]].query(
    f"Dataset == '{dataset}'"
)
dataset_sub.set_index("Method", inplace=True)
fig = plt.figure(figsize=(3, 1), dpi=400)

ax = fig.gca()
dataset_sub = dataset_sub.reindex(names)
sim = dataset_sub[["Cosine sim. orig"]].values.flatten()
errors = dataset_sub[["Cosine sim. 95%"]].values.flatten()
method_names = dataset_sub.index.values

valid_indices = np.isfinite(sim) & (sim > 0)

x_pos = np.arange(valid_indices.sum())
ax.bar(x_pos, 
       sim[valid_indices], 
       color=np.array([method_colors[i] for i in method_names])[valid_indices], 
)
metric_points = get_metric_points(active_adduct, dataset, "cos_sim")
plotted_methods = np.array(method_names)[valid_indices]
n_counts = {method: len(metric_points.get(method, [])) for method in plotted_methods}
overlay_metric_points(ax, method_names, valid_indices, metric_points)
draw_visible_ci(ax, x_pos, sim[valid_indices], errors[valid_indices])
annotate_n_counts(ax, x_pos, sim[valid_indices], plotted_methods, n_counts)

# Add 95% conf bars


ax.set_ylabel("Cosine sim. avg")
ax.set_xlabel("Method")
# Rotate x tick labels
ax.set_xticks(x_pos)
ax.set_xticklabels(plotted_methods, rotation=45, ha="right")
```

```python
def plot_cosine_retrieval_for_adduct(active_adduct):
    out_df = adduct_to_df[active_adduct]

    # Plot cosine similarities and time for 2 datasets
    gridspec = dict(hspace=0, width_ratios=[1, 1.5, 0.3, 1, 1.5])
    fig, axes = plt.subplots(
        1, 5, figsize=(8.6, 1.4), dpi=400, sharey=False, gridspec_kw=gridspec
    )
    plt.subplots_adjust(wspace=0.1, hspace=0.1)
    plt_dataset_names = {"split_1_nist20": "InChIKey", "scaffold_1_nist20": "Scaffold"}

    plot_datasets = [
        "split_1_nist20",
        "scaffold_1_nist20",
    ]
    spectral_n_counts = {}
    for ind, (cur_axes, dataset) in enumerate(zip([axes[0:2], axes[3:5]], plot_datasets)):
        # Cosine similarity plot
        ax = cur_axes[0]

        dataset_sub = out_df[
            ["Method", "Cosine sim. orig", "Cosine sim. 95%", "Dataset"]
        ].query(f"Dataset == '{dataset}'")
        dataset_sub.set_index("Method", inplace=True)
        dataset_sub = dataset_sub.reindex(names)
        sim = dataset_sub[["Cosine sim. orig"]].values.flatten()
        ci = dataset_sub[["Cosine sim. 95%"]].values.flatten()

        # Cosine sim. 95% is already the 95% confidence interval.

        method_names = dataset_sub.index.values
        valid_indices = np.isfinite(sim) & (sim > 0)
        plotted_methods = np.array(method_names)[valid_indices]
        x_pos = np.arange(len(plotted_methods))

        ax.bar(
            x_pos,
            sim[valid_indices],
            color=np.array([method_colors[i] for i in method_names])[valid_indices],
            edgecolor="black",
            linewidth=0.5,
        )
        metric_points = get_metric_points(active_adduct, dataset, "cos_sim")
        overlay_metric_points(ax, method_names, valid_indices, metric_points)
        draw_visible_ci(ax, x_pos, sim[valid_indices], ci[valid_indices])
        n_counts = {method: len(metric_points.get(method, [])) for method in plotted_methods}
        spectral_n_counts[dataset] = n_counts
        annotate_n_counts(ax, x_pos, sim[valid_indices], plotted_methods, n_counts)
        adduct_display = adduct_to_display.get(active_adduct, f"of {active_adduct}")
        ax.set_title(f"Cosine similarity {adduct_display}")
        ax.set_xticks(x_pos)
        ax.set_xticklabels(
            plotted_methods,
            rotation=45,
            ha="right",
            rotation_mode="anchor",
            fontsize=6,
        )
        dx = 5/72.; dy = 2/72.
        offset = matplotlib.transforms.ScaledTranslation(dx, dy, fig.dpi_scale_trans)
        for label in ax.xaxis.get_majorticklabels():
            label.set_transform(label.get_transform() + offset)
        ax.yaxis.labelpad = 1

        # Retrieval acc plot
        ax = cur_axes[1]
        for method in names:
            sub_df = out_df[
                (out_df["Method"] == method) & (out_df["Dataset"] == dataset)
            ]
            if sub_df.empty:
                continue
            topk, accs, error = [0], [0], [0]
            for i in range(1, 11):
                acc = sub_df[f"Top_{i} Retrieval"].item()
                err = sub_df[f"Top_{i} Retrieval 95%"].item()
                if acc != 0:
                    topk.append(i)
                    accs.append(acc)
                    if err != 0:
                        error.append(err)
            marker = '.'
            if len(accs) > 1:
                ax.plot(
                    topk,
                    accs,
                    color=method_colors[method],
                    linewidth=0.5,
                    label=method,
                    marker=marker,
                    markersize=0.8,
                )
                ax.legend(fontsize=5)
                ax.fill_between(
                    topk,
                    np.zeros_like(accs),
                    np.array(accs),
                    alpha=0.05,
                    color=method_colors[method],
                )

            if len(accs) > 1 and method in iceberg_ours_names + iceberg_legacy_names:
                if method in iceberg_legacy_names and dataset == 'split_1_nist20':
                    ax.text(topk[1]+0.2, accs[1]-0.01, f'{accs[1]:.2f}',
                            horizontalalignment='left', verticalalignment='top', fontsize=5, color=method_colors[method])
                else:
                    ax.text(topk[1]-0.25, accs[1]+0.01, f'{accs[1]:.2f}',
                            horizontalalignment='right', fontsize=5, color=method_colors[method])
                ax.text(topk[-1]+0.7, accs[-1], f'{accs[-1]:.2f}',
                        horizontalalignment='left', verticalalignment='center', fontsize=5, color=method_colors[method])

        ax.xaxis.set_ticks(np.arange(1, 11))
        ax.set_xlabel("Top-k", fontfamily='arial')
        ax.set_title(f"Retrieval accuracy {adduct_display}")

    axes[0].sharey(axes[1])
    axes[0].set_ylim(0, 1)
    axes[1].set_xticklabels(list(range(1, 11)))
    axes[1].tick_params(labelleft=False)
    axes[2].axis("off")
    axes[3].sharey(axes[4])
    axes[3].set_ylim(0, 1)
    axes[4].set_xticklabels(list(range(1, 11)))
    axes[4].tick_params(labelleft=False)

    out_path = results_folder / f"cosine_retrieval_{active_adduct}.pdf"
    fig.savefig(out_path, bbox_inches="tight", transparent=True)
    if active_adduct == "avg":
        fig.savefig(results_folder / "cosine_retrieval.pdf", bbox_inches="tight", transparent=True)
    plt.close(fig)


for active_adduct in ['[M+H]+', '[M-H]-', 'pos_avg', 'neg_avg', 'avg', 'metfrag']:
    plot_cosine_retrieval_for_adduct(active_adduct)

active_adduct = '[M+H]+'
out_df = adduct_to_df[active_adduct]
```

```python
import contextlib
from ms_pred import common

ICEBERG20_COMPAT_ION2ONEHOT_POS = {
    "[M+H]+": 0,
    "[M+Na]+": 1,
    "[M+K]+": 2,
    "[M-H2O+H]+": 3,
    "[M+H-H2O]+": 3,
    "[M+H3N+H]+": 4,
    "[M+NH4]+": 4,
    "[M]+": 5,
    "[M-H4O2+H]+": 6,
    "[M+H-2H2O]+": 6,
    "[M-H]-": 7,
    "[M+Cl]-": 8,
    "[M-H2O-H]-": 9,
    "[M-H-H2O]-": 9,
    "[M-H-CO2]-": 10,
}


@contextlib.contextmanager
def iceberg20_adduct_compat():
    old_ion2onehot_pos = common.ion2onehot_pos
    old_onehot_pos2ion = common.onehot_pos2ion
    try:
        common.ion2onehot_pos = ICEBERG20_COMPAT_ION2ONEHOT_POS.copy()
        common.onehot_pos2ion = {v: k for k, v in common.ion2onehot_pos.items()}
        yield
    finally:
        common.ion2onehot_pos = old_ion2onehot_pos
        common.onehot_pos2ion = old_onehot_pos2ion

def load_yaml_to_df(yaml_path: str) -> pd.DataFrame:
    with open(yaml_path, "r") as f:
        data = yaml.safe_load(f)

    individuals = data.get("individuals", [])
    rows = []

    for ind in individuals:
        row = {
            "ind_recovered": ind.get("ind_recovered", np.nan),
            "mass": ind.get("mass", np.nan),
            "num_peaks_avg": ind.get("num_peaks_avg", np.nan),
            "num_collision_engs": ind.get("num_collision_engs", np.nan),
            "ion": ind.get("ion", None),
        }
        for i in range(1, 51):
            row[f"top_{i}_dist"] = ind.get(f"top_{i}_dist", np.nan)
        if row["ion"] == '[M+H]+' or row["ion"] == '[M-H]-':
            rows.append(row)

    df = pd.DataFrame(rows)

    # Enforce numeric where expected
    for c in ["ind_recovered", "mass", "num_peaks_avg", "num_collision_engs"] + [f"top_{i}_dist" for i in range(1, 51)]:
        df[c] = pd.to_numeric(df[c], errors="coerce")

    return df


# ----------------------------
# Feature engineering
# ----------------------------
def ion_onehot(ion: str) -> np.ndarray:
    """One-hot with all-zeros fallback for unknown/None ions."""
    n_ions = max(common.ion2onehot_pos.values()) + 1
    vec = np.zeros(n_ions, dtype=np.float32)
    if ion is None:
        return vec
    idx = common.ion2onehot_pos.get(ion)
    if idx is None:
        return vec
    vec[idx] = 1.0
    return vec


def build_feature_matrix(df: pd.DataFrame, topk_features) -> np.ndarray:
    dist_cols = [f"top_{i}_dist" for i in range(1, topk_features + 1)]
    num_cols = dist_cols + ["mass", "num_peaks_avg", "num_collision_engs"]

    X_num = df[num_cols].to_numpy(dtype=np.float32)
    X_ion = np.stack([ion_onehot(x) for x in df["ion"].tolist()], axis=0).astype(np.float32)

    return np.concatenate([X_num, X_ion], axis=1)


def build_label(df: pd.DataFrame, k: int) -> np.ndarray:
    r = df["ind_recovered"].to_numpy(dtype=np.float32)
    # label 1 if recovered within top-k
    return (r <= float(k)).astype(np.int32)
```

```python
def coverage_accuracy_curve(p, y, grid=None):
    """
    For thresholds t: keep items with p>=t.
    Returns coverage (kept fraction) and accuracy among kept.
    """
    p = np.asarray(p)
    y = np.asarray(y).astype(int)
    if grid is None:
        grid = np.linspace(0, 1, 201)

    coverage = []
    acc = []
    for t in grid:
        m = (p >= t)
        if m.sum() == 0:
            coverage.append(0.0)
            acc.append(np.nan)
        else:
            coverage.append(m.mean())
            acc.append(y[m].mean())  # fraction positive among kept
    return grid, np.array(coverage), np.array(acc)

def pick_threshold_for_target_acc(p, y, target_acc, min_coverage=0.02, grid=None):
    """
    Small helper: pick the *lowest* threshold achieving target accuracy
    (to maximize coverage), subject to min_coverage.
    """
    t, cov, acc = coverage_accuracy_curve(p, y, grid=grid)
    ok = (cov >= min_coverage) & np.isfinite(acc) & (acc >= target_acc)
    if not np.any(ok):
        return None
    return t[ok][0]  # lowest t that meets target
```

```python
# Joint, paste-ready notebook cell:
# - Computes ROC (your original)
# - Adds reliability diagram (p vs empirical accuracy)
# - Adds coverage–accuracy curve (threshold tradeoff)
# - Derives "low/mid/high confidence" p-ranges from target accuracy levels
#

import numpy as np
import pandas as pd
import joblib
import matplotlib.pyplot as plt
from pathlib import Path
from sklearn.metrics import roc_curve, roc_auc_score

# -------------------------
# Config (edit as needed)
# -------------------------
conf_model_dir = Path('/home/runzhong/ms-models/iceberg_results_20260413/iceberg_nist20/split_1_rnd1/retrieval_confidence').expanduser()
test_yaml = Path("/home/runzhong/ms-models/iceberg_results_20260413/iceberg_nist20/split_1_rnd1/retrieval_nist20_split_1_50/rerank_eval_entropy.yaml")

results_folder = Path("results/diagnostics_confidence")  # change if desired
results_folder.mkdir(parents=True, exist_ok=True)

KS = [1, 3, 5, 10]
TOPK_FEATURES = 15  # must match training

COLORS = {
    1: '#7B94CC',
    3: '#E9B382',
    5: '#FFD593',
    10: '#7C9D97',
}

# Confidence band targets:
# choose what "high/mid" should mean in terms of empirical accuracy among kept (precision)
TARGET_ACC_HIGH = 0.95
TARGET_ACC_MID  = 0.85
MIN_COVERAGE_HIGH = 0.00
MIN_COVERAGE_MID  = 0.02

# Binning for reliability plot (quantile bins = equal counts; more stable)
RELIABILITY_BINS = 10
RELIABILITY_STRATEGY = "quantile"  # or "uniform"

# -------------------------
# Helper functions
# -------------------------
def binned_stats(p, y, bins=10, strategy="quantile"):
    """
    Per-bin mapping from p -> empirical accuracy.
    Returns a dataframe with n, mean_p, emp_acc, p_min, p_max, and approx 95% CI.
    """
    df = pd.DataFrame({"p": np.asarray(p), "y": np.asarray(y).astype(int)}).sort_values("p")

    if strategy == "quantile":
        df["bin"] = pd.qcut(df["p"], q=bins, duplicates="drop")
    elif strategy == "uniform":
        df["bin"] = pd.cut(df["p"], bins=np.linspace(0, 1, bins + 1), include_lowest=True)
    else:
        raise ValueError("strategy must be 'quantile' or 'uniform'")

    g = df.groupby("bin", observed=True)
    out = g.agg(
        n=("y", "size"),
        mean_p=("p", "mean"),
        emp_acc=("y", "mean"),
        p_min=("p", "min"),
        p_max=("p", "max"),
    ).reset_index(drop=True)

    # Simple normal-approx CI for display
    se = np.sqrt(np.clip(out["emp_acc"] * (1 - out["emp_acc"]) / out["n"], 0, None))
    out["emp_acc_lo"] = np.clip(out["emp_acc"] - 1.96 * se, 0, 1)
    out["emp_acc_hi"] = np.clip(out["emp_acc"] + 1.96 * se, 0, 1)
    return out

def plot_reliability(ax, stats, title):
    ax.plot([0, 1], [0, 1], linestyle="--", linewidth=0.8, color="#666666")
    ax.errorbar(
        stats["mean_p"], stats["emp_acc"],
        yerr=[stats["emp_acc"] - stats["emp_acc_lo"], stats["emp_acc_hi"] - stats["emp_acc"]],
        fmt="o", markersize=2.5, linewidth=0.8, capsize=2
    )
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.set_xlabel("Mean predicted p (bin)")
    ax.set_ylabel("Empirical accuracy (fraction positive)")
    ax.set_title(title)

def coverage_accuracy_curve(p, y, grid=None):
    """
    For thresholds t: keep items with p>=t.
    Returns:
      t_grid, coverage (kept fraction), acc_kept (mean(y) among kept).
    """
    p = np.asarray(p)
    y = np.asarray(y).astype(int)

    if grid is None:
        grid = np.linspace(0, 1, 401)

    coverage = np.zeros_like(grid, dtype=float)
    acc_kept = np.full_like(grid, np.nan, dtype=float)

    for i, t in enumerate(grid):
        m = (p >= t)
        coverage[i] = m.mean()
        if m.sum() > 0:
            acc_kept[i] = y[m].mean()
    return grid, coverage, acc_kept

def pick_threshold_for_target_acc(p, y, target_acc, min_coverage=0.02, grid=None):
    """
    Pick the lowest threshold achieving acc_kept >= target_acc
    while keeping coverage >= min_coverage.
    Returns None if impossible.
    """
    t, cov, acc = coverage_accuracy_curve(p, y, grid=grid)
    ok = (cov >= min_coverage) & np.isfinite(acc) & (acc >= target_acc)
    if not np.any(ok):
        return None
    return float(t[ok][0])

def summarize_bands(p, y, t_mid, t_high):
    """
    Build a dataframe summarizing low/mid/high bands with empirical accuracy and coverage.
    """
    p = np.asarray(p)
    y = np.asarray(y).astype(int)

    if (t_mid is None) and (t_high is None):
        band = np.array(["low"] * len(p))
    else:
        band = np.full(len(p), "low", dtype=object)
        if t_mid is not None:
            band[p >= t_mid] = "mid"
        if t_high is not None:
            band[p >= t_high] = "high"

    df = pd.DataFrame({"p": p, "y": y, "band": band})
    out = df.groupby("band").agg(
        n=("y", "size"),
        emp_acc=("y", "mean"),
        p_min=("p", "min"),
        p_max=("p", "max"),
    ).reset_index()
    out["coverage"] = out["n"] / len(df)

    order = {"high": 0, "mid": 1, "low": 2}
    out["order"] = out["band"].map(order).fillna(99).astype(int)
    out = out.sort_values("order").drop(columns=["order"])
    return out

# -------------------------
# Load confidence models and match the adduct feature block to their weights
# -------------------------
def load_conf_pipeline(k):
    model_path = conf_model_dir / f"lr_conf_k{k}.joblib"
    obj = joblib.load(model_path)
    return obj["pipeline"] if isinstance(obj, dict) and "pipeline" in obj else obj


def get_expected_feature_count(pipe):
    expected = getattr(pipe, "n_features_in_", None)
    if expected is not None:
        return expected
    if hasattr(pipe, "named_steps"):
        for step in pipe.named_steps.values():
            expected = getattr(step, "n_features_in_", None)
            if expected is not None:
                return expected
    return None


conf_pipes = {k: load_conf_pipeline(k) for k in KS}
expected_feature_counts = {
    get_expected_feature_count(pipe)
    for pipe in conf_pipes.values()
    if get_expected_feature_count(pipe) is not None
}
if len(expected_feature_counts) > 1:
    raise ValueError(f"Confidence predictors disagree on feature counts: {sorted(expected_feature_counts)}")
expected_features = next(iter(expected_feature_counts), None)

numeric_feature_count = TOPK_FEATURES + 3
current_adduct_features = max(common.ion2onehot_pos.values()) + 1
iceberg20_adduct_features = max(ICEBERG20_COMPAT_ION2ONEHOT_POS.values()) + 1
adduct_context = contextlib.nullcontext()
adduct_mapping_name = "current chem_utils"
if expected_features == numeric_feature_count + iceberg20_adduct_features:
    adduct_context = iceberg20_adduct_compat()
    adduct_mapping_name = "ICEBERG 2.0 compatibility"
elif expected_features not in (None, numeric_feature_count + current_adduct_features):
    raise ValueError(
        f"Confidence predictor expects {expected_features} features, but neither the current "
        f"adduct mapping ({numeric_feature_count + current_adduct_features}) nor the ICEBERG 2.0 "
        f"compatibility mapping ({numeric_feature_count + iceberg20_adduct_features}) matches."
    )

with adduct_context:
    test_df = load_yaml_to_df(test_yaml)
    required = ["ind_recovered", "mass", "num_peaks_avg"]
    test_df = test_df.dropna(subset=required).copy()
    X_test = build_feature_matrix(test_df, topk_features=TOPK_FEATURES)

print(f"Confidence predictor features: {X_test.shape[1]} columns using {adduct_mapping_name} adduct mapping")

# -------------------------
# 1) ROC plot (all k)
# -------------------------
fig_roc, ax_roc = plt.subplots(1, 1, figsize=(2.0, 2.0), dpi=400)
ax_roc.plot([0, 1], [0, 1], linestyle="--", linewidth=0.6, color="#666666")

threshold_rows = []  # collect thresholds + summary per k

for k in KS:
    pipe = conf_pipes[k]

    y_test = build_label(test_df, k=k).astype(int)
    if len(np.unique(y_test)) < 2:
        print(f"[k={k}] skip: y_test has one class (pos={y_test.sum()}/{len(y_test)})")
        continue

    expected_features = get_expected_feature_count(pipe)
    if expected_features is not None and X_test.shape[1] != expected_features:
        raise ValueError(
            f"Confidence predictor expects {expected_features} features, "
            f"but the notebook built {X_test.shape[1]}. Check the ICEBERG 2.0 "
            "adduct compatibility mapping before prediction."
        )

    p_test = pipe.predict_proba(X_test)[:, 1]

    # ROC
    fpr, tpr, _ = roc_curve(y_test, p_test)
    auc = roc_auc_score(y_test, p_test)
    ax_roc.plot(fpr[::2], tpr[::2], linewidth=0.6, label=f"k={k} (AUC={auc:.3f})", color=COLORS.get(k, None))

    # -------------------------
    # 2) Reliability plot (per k)
    # -------------------------
    stats = binned_stats(p_test, y_test, bins=RELIABILITY_BINS, strategy=RELIABILITY_STRATEGY)

    fig_rel, ax_rel = plt.subplots(1, 1, figsize=(2.2, 2.0), dpi=400)
    plot_reliability(ax_rel, stats, title=f"Reliability: P(true in top-{k}) vs p")
    fig_rel.tight_layout()
    fig_rel.savefig(results_folder / f"reliability_k{k}.pdf", bbox_inches="tight", transparent=True)
    plt.close(fig_rel)

    # Save binned table
    stats.to_csv(results_folder / f"reliability_bins_k{k}.csv", index=False)

    # -------------------------
    # 3) Coverage–accuracy plot + threshold selection
    # -------------------------
    grid, cov, acc_kept = coverage_accuracy_curve(p_test, y_test)

    t_high = pick_threshold_for_target_acc(
        p_test, y_test, target_acc=TARGET_ACC_HIGH, min_coverage=MIN_COVERAGE_HIGH, grid=grid
    )
    t_mid = pick_threshold_for_target_acc(
        p_test, y_test, target_acc=TARGET_ACC_MID, min_coverage=MIN_COVERAGE_MID, grid=grid
    )

    fig_cov, ax_cov = plt.subplots(1, 1, figsize=(2.2, 2.0), dpi=400)
    ax_cov.plot(cov, acc_kept, linewidth=0.9)
    ax_cov.set_xlim(0, 1)
    ax_cov.set_ylim(0, 1)
    ax_cov.set_xlabel("Coverage (fraction kept with p ≥ t)")
    ax_cov.set_ylabel(f"Empirical accuracy among kept\n(P(true in top-{k}) | kept)")
    ax_cov.set_title(f"Coverage–Accuracy (k={k})")

    # Mark chosen thresholds on the curve (if they exist)
    def mark_threshold(t_mark, label):
        if t_mark is None:
            return
        idx = np.argmin(np.abs(grid - t_mark))
        ax_cov.plot([cov[idx]], [acc_kept[idx]], marker="o", markersize=3)
        ax_cov.text(cov[idx], acc_kept[idx], f" {label}: t={t_mark:.2f}", fontsize=6, va="bottom")

    mark_threshold(t_mid, "mid")
    mark_threshold(t_high, "high")

    fig_cov.tight_layout()
    fig_cov.savefig(results_folder / f"coverage_accuracy_k{k}.pdf", bbox_inches="tight", transparent=True)
    plt.close(fig_cov)

    # -------------------------
    # 4) Band summary (low/mid/high)
    # -------------------------
    band_summary = summarize_bands(p_test, y_test, t_mid=t_mid, t_high=t_high)
    band_summary.to_csv(results_folder / f"confidence_bands_k{k}.csv", index=False)

    # Collect a compact per-k summary row
    threshold_rows.append({
        "k": k,
        "auc": auc,
        "base_rate_pos": float(y_test.mean()),
        "t_mid_for_acc>=%.2f" % TARGET_ACC_MID: t_mid,
        "t_high_for_acc>=%.2f" % TARGET_ACC_HIGH: t_high,
    })

# finalize ROC
ax_roc.set_xlim(0, 1)
ax_roc.set_ylim(0, 1)
ax_roc.set_xlabel("False Positive Rate")
ax_roc.set_ylabel("True Positive Rate")
ax_roc.set_title("True structure in top-k predictions (NIST'20)")
ax_roc.legend(fontsize=5, loc="lower right")
fig_roc.tight_layout()
fig_roc.savefig(results_folder / "roc_k1_3_5_10.pdf", bbox_inches="tight", transparent=True)

# -------------------------
# 5) Save threshold summary table
# -------------------------
threshold_df = pd.DataFrame(threshold_rows).sort_values("k")
threshold_df.to_csv(results_folder / "confidence_thresholds_summary.csv", index=False)

display(threshold_df)

print(f"\nSaved diagnostics to: {results_folder.resolve()}")
print("Files per k:")
print(" - reliability_k{k}.pdf + reliability_bins_k{k}.csv")
print(" - coverage_accuracy_k{k}.pdf")
print(" - confidence_bands_k{k}.csv")
print("Plus: roc_k1_3_5_10.pdf and confidence_thresholds_summary.csv")
```

```python
mass_to_df = {}
for mass_label in ['(0, 5]', '(5, 10]', '(10, 15]', '(15, 20]', '(20, 25]', '(25, 30]', '(30, 40]', '(40, 500]']:
    out_df = []
    for dataset_name, tsv_files in dataset_to_res.items():
        for k, v in tsv_files.items():
            interested_rows = [vv[vv.iloc[:, 0] == mass_label] for vv in v if vv is not None]
            if not interested_rows or any([len(row) == 0 for row in interested_rows]):
                continue
            def get_val(key):
                vals = []
                for row in interested_rows:
                    if key in row:
                        vals.append(row[key].item())
                return vals
            new_entry = {
                "Method": k,
                "Cosine sim.": np.mean(get_val('cos_sim')),
                "Cosine sim. 95%": 1.96 * sem(get_val('cos_sim')),

                "Cosine sim. (no PEP)": np.mean(get_val('cos_sim_zero_pep')),
                "Cosine sim. 95% (no PEP)": 1.96 * sem(get_val('cos_sim_zero_pep')),

                "Entropy sim.": np.mean(get_val('entropy_sim')),
                "Entropy sim. 95%": 1.96 * sem(get_val('entropy_sim')),

                "Coverage": np.mean(get_val('coverage')),
                "Coverage 95%": 1.96 * sem(get_val('coverage')),

                # The way of computing validity in this code base is outdated
                #"Valid": np.mean(get_val('frac_valid')),
                #"Valid 95%": 1.96 * sem(get_val('frac_valid')),

                "Dataset": dataset_name,
            }
            new_entry.update({
                f"Top_{i} Retrieval": np.mean(get_val(f"top_{i}")) for i in range(1, 11)
            })
            new_entry.update({
                f"Top_{i} Retrieval 95%": 1.96 * sem(get_val(f"top_{i}")) for i in range(1, 11)
            })
            out_df.append(new_entry)
    out_df = pd.DataFrame(out_df)

    # Replace nan with 0 
    out_df = out_df.fillna(0)
    
    mass_to_df[mass_label] = out_df
```
